```python
import csv

def transform_file(source_path, dest_path):
    # Open the source text file and destination CSV file.
    with open(source_path, 'r') as infile, open(dest_path, 'w', newline='') as outfile:
        writer = csv.writer(outfile)
        # If you need a header, uncomment the next line:
        # writer.writerow(["ucb_iteration", "idx", "ucb_value", "importance_value", "n_idx", "n_total"])
        for line in infile:
            line = line.strip()
            if not line:
                continue  # Skip empty lines
            # Expected line format: "idx: <number>, pixel importance: <value>"
            try:
                parts = line.split(',')
                # Extract idx (removing "idx:" and extra spaces)
                idx = parts[0].split(':')[1].strip()
                # Extract pixel importance value (removing "pixel importance:" and extra spaces)
                importance = parts[1].split(':')[1].strip()
            except IndexError:
                print(f"Skipping malformed line: {line}")
                continue
            # Build the row: fixed first column "0", then idx, constant ucb_value "0.8224693",
            # then the pixel importance value, then fixed "1" and "50176"
            row = ["0", idx, "0.8224693", importance, "1", "50176"]
            writer.writerow(row)


source_file = "pixel_info_Horse_Zebra_Zebra.txt"     # update with your source file path
dest_file = "Horse_Zebra/Zebra/ucb_log.csv"        # update with your destination file path
transform_file(source_file, dest_file)
print("Transformation complete!")
```
